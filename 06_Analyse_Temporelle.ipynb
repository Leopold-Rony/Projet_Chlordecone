# Analyse Temporelle de la Contamination

## Évolution du taux de chlordécone au fil des années
Afin de compléter notre diagnostic spatial et multivarié, il est crucial d'étudier la dimension temporelle de notre jeu de données. La chlordécone étant une molécule très persistante dans l'environnement, nous allons vérifier si le taux moyen mesuré sur les parcelles évolue entre les premières campagnes de prélèvement (2003) et les plus récentes (2019).

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# 1. Chargement des données propres (si tu es dans un nouveau Notebook)
df = pd.read_csv('chlordecone_clean.csv')
```

```python
# 2. Calcul du taux moyen de chlordécone par année
evolution_annuelle = df.groupby('ANNEE')['Taux_Chlordecone'].mean().reset_index()
```

```python
# 3. Création du graphique (Line plot)
plt.figure(figsize=(12, 6))
sns.lineplot(
    data=evolution_annuelle, 
    x='ANNEE', 
    y='Taux_Chlordecone', 
    marker='o',      # Ajoute un point pour chaque année
    color='firebrick', 
    linewidth=2.5
)

# 4. Esthétique du graphique
plt.title('Évolution du taux moyen de Chlordécone détecté par année (2003-2019)', fontsize=16, fontweight='bold')
plt.xlabel('Année de prélèvement', fontsize=12)
plt.ylabel('Taux moyen de Chlordécone', fontsize=12)

# Forcer l'affichage de toutes les années sur l'axe X (en entier)
plt.xticks(evolution_annuelle['ANNEE'], rotation=45)
plt.grid(True, linestyle='--', alpha=0.7)

# On force l'axe Y à commencer à 0 pour ne pas fausser la perception visuelle
plt.ylim(bottom=0) 

plt.tight_layout()
plt.show()
```

### 2. Interprétation de la tendance temporelle
L'analyse de la courbe révèle une baisse apparente du taux moyen de chlordécone détecté après 2010 (stagnation autour de 0.5 µg/kg), succédant à des taux plus élevés dans les années 2000. 

D'un point de vue agronomique, la chlordécone étant une molécule extrêmement persistante (demi-vie de plusieurs siècles), cette baisse ne reflète pas une dépollution des sols. Elle illustre très probablement un **élargissement des campagnes de prélèvements** : les premières analyses ciblaient en priorité les zones historiquement contaminées (bananeraies), tandis que les campagnes récentes couvrent un territoire plus vaste, incluant des zones saines qui "diluent" la moyenne globale.
